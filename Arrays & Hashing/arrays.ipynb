# Arrays in Python

## Properities
- Conitguous allocation of memory
- Can contain homogeneous and heterogeneous both types unlike C, C++, Java 
- Arrays are by default dynamic in python. We can use it without allocating its size.




## How to use Arrays in Python
- Use Lists - Most preferred 
- Use Python's array module. 
> Syntax -> array.array(<TypeCode>, <elements>) 
- Use Numpy


```python
# Array Module

from array import *

val = array('i',[1,2,3,4,5,6])

#print complete list or iterate
print(val)

#print typecode
print(val.typecode)

#reverse array
val.reverse()
print(val)

#insert elements
val.insert(1,50)
print(val)

#insert at the end
val.append(200)
print(val)

#copy array
copyArray = array(val.typecode,  (x for x in val))
print(copyArray)
```

```python
def likes(names):
    # your code here
    like_msg=" likes this"
    if len(names)==0:
        print("no one"+ like_msg)
    elif len(names)==1:
        print(''.join(map(str,names)) + like_msg)
    elif len(names)==2:
        print(' and '.join(map(str,names)) + like_msg)
    elif len(names)==3:
        print(','.join(map(str,names[:2])) + ' and '+ str(names[2])+ like_msg)
    elif len(names)>3:
        print(','.join(map(str,names[:2])) + ' and '+ str(len(names[2:])) + "others " + like_msg)

likes(['jacob', 'james'])
```

```python
def narcissistic(value):
    digit_sum=0
    value_len= len(str(value))
    print(value_len)
    for digit in str(value):
        digit_sum += pow(int(digit),value_len)
        print(digit_sum)
    
    if digit_sum==value:
        return True
    return False

narcissistic(122)
```
