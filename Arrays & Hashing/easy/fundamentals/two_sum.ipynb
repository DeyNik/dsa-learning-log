## Problem Statement

Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

You may assume that each input would have **exactly one solution**, and you may not use the same element twice. You can return the answer in any order.

---

## Examples

### Example 1:

* **Input:** `nums = [2,7,11,15]`, `target = 9`
* **Output:** `[0,1]`
* **Explanation:** `nums[0] + nums[1] == 2 + 7 == 9`, so we return `[0, 1]`.

### Example 2:

* **Input:** `nums = [3,2,4]`, `target = 6`
* **Output:** `[1,2]`

### Example 3:

* **Input:** `nums = [3,3]`, `target = 6`
* **Output:** `[0,1]`

---

## Constraints

* $2 \le \text{nums.length} \le 10^4$
* $-10^9 \le \text{nums}[i] \le 10^9$
* $-10^9 \le \text{target} \le 10^9$
* Only one valid answer exists.

---

## Follow-up: Less than $\mathcal{O}(n^2)$ Time Complexity?

Remember our discussion about **Hash Maps (`dict`)**? This is the classic problem where a hash map transforms a slow $\mathcal{O}(n^2)$ brute-force solution (using nested loops) into a lightning-fast **$\mathcal{O}(n)$** solution.

### The Intuition

As you iterate through the array, for every number, you can calculate its **complement**:


$$\text{complement} = \text{target} - \text{current\_number}$$

Instead of looking ahead or searching the rest of the array, ask yourself: *"Have I seen this complement before?"*

* If **yes**, you've found your pair! Return its stored index and your current index.
* If **no**, store the current number and its index in your Hash Map (`dict`) and move on.



```python
def twoSum(nums: list[int], target: int) :
        hash_map = {}

        for index,item in enumerate(nums):

            if (target - item) in hash_map:
                return [index, hash_map[target-item] ]

            hash_map[item]=index

            
print(twoSum([2,7,11,15],9))            
```

## Two Sum

*Two Sum* is a classic programming interview problem and the first problem on LeetCode. It asks for the indices of two numbers in an array whose values add up to a given target, making it a foundational exercise in **hash table** usage and algorithmic problem solving. Its popularity comes from illustrating how a simple change in data structure can improve performance from quadratic to linear time.







### Problem statement

Given an array of integers and a target value, the task is to find the **indices** of the two distinct elements whose sum equals the target. The standard version guarantees that exactly one valid answer exists, and the same array element cannot be used twice. The emphasis is on returning positions rather than the values themselves.

### Efficient solution

The most common approach uses a **hash map** that stores previously seen numbers and their indices. As each new value is processed, the algorithm computes the required complement (*target − current value*) and checks whether it has already been seen. This produces an **O(n)** time solution with **O(n)** additional space, which is substantially faster than checking every possible pair.

### Why it matters

Although the problem is straightforward, it introduces several important algorithmic ideas: choosing the right data structure, trading memory for speed, and recognizing lookup patterns. These concepts appear repeatedly in interview questions involving arrays, maps, strings, and graph algorithms, making Two Sum a common starting point for learning coding interview techniques.

### Common pitfalls

Frequent mistakes include returning the values instead of their indices, reusing the same element twice, mishandling duplicate numbers, or using a nested-loop solution when an optimal linear-time approach is expected. Carefully tracking indices while updating the hash map avoids these issues and leads to a clean, efficient implementation.

### How does hash maps work?

```python
hash_map={}

for i in range(2,8):
    hash_map[i]=i*2

print(hash_map)
```
