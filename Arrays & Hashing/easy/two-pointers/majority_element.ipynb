### Problem Description

Given an array `nums` of size `n`, return the majority element.

The majority element is the element that appears more than $\lfloor n / 2 \rfloor$ times. You may assume that the majority element always exists in the array.

---

### Examples

**Example 1:**

* **Input:** `nums = [3, 2, 3]`
* **Output:** `3`

**Example 2:**

* **Input:** `nums = [2, 2, 1, 1, 1, 2, 2]`
* **Output:** `2`

---

### Constraints

* $n == \text{nums.length}$
* $1 \le n \le 5 \times 10^4$
* $-10^9 \le \text{nums}[i] \le 10^9$
* The input is generated such that a majority element will exist in the array.

---

### Follow-up

Could you solve the problem in linear time and in $O(1)$ space?

````python
#Takes O(n) space 
def majority_element(nums):
    hash_map={}
    for i in range(len(nums)):
        if nums[i] in hash_map:
            hash_map[nums[i]]+=1
            continue
        hash_map[nums[i]]= 1

    print(hash_map)

    return max(hash_map, key=hash_map.get)

majority_element([3,2,3])
````

## Boyer-Moore Voting Algorithm.

Yes! The optimal approach for solving the Majority Element problem in **$O(n)$ time** and **$O(1)$ space** is the **Boyer-Moore Voting Algorithm**.

### How It Works

1. Maintain a `candidate` and a `count`.
2. Iterate through the array:
* If `count` is `0`, assign the current element as the new `candidate`.
* If the current element matches the `candidate`, increment `count`. Otherwise, decrement `count`.


3. Because the majority element appears more than $\lfloor n / 2 \rfloor$ times, it will always outvote all other elements combined and remain as the final candidate.

### Python Implementation


````python
def majority_element(nums):
    candidate=0
    count=0
    for num in nums:
        if count==0:
            candidate=num
        count += (1 if num==candidate else -1)
        print(count, candidate)

    
    print("candidate",candidate, "count", count)

import math
print("floor n/2 is", math.floor(3/2))
majority_element([3,2,3])
print("floor n/2 is", math.floor(7/2))
majority_element([2, 2, 1, 1, 1, 2, 2])
print("floor n/2 is", math.floor(3/2))
majority_element([6,5,5])
````

### Core idea

The optimal solution is the **Boyer–Moore Majority Vote algorithm**. It keeps a `candidate` and a `count`: when `count` reaches zero, the current number becomes the new candidate; matching values increment the count, while different values decrement it. This achieves the requested **O(n) time and O(1) space**. [LeetCode](https://leetcode.com/problems/majority-element/?envId=eeudwo2i\&envType=problem-list-v2\&utm_source=chatgpt.com)

### Why Boyer–Moore works

Think of every occurrence of the majority element being **canceled out** by a different element. Because the majority appears more than half the time, there cannot be enough non-majority elements to cancel every occurrence. Consequently, after all possible cancellations, the surviving candidate must be the majority element—given LeetCode 169's guarantee that one exists.

### Example

For `[2, 2, 1, 1, 1, 2, 2]`, the candidate's count rises when another `2` appears and falls when a different value appears. Although the count can return to zero and the candidate can change during the scan, the final candidate is `2`, which occurs four times out of seven. [LeetCode](https://leetcode.com/problems/majority-element/?envId=eeudwo2i\&envType=problem-list-v2\&utm_source=chatgpt.com)

### Typical implementation

```python
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        candidate = None
        count = 0

        for num in nums:
            if count == 0:
                candidate = num

            count += 1 if num == candidate else -1

        return candidate
```

The key interview insight is that a hash map also gives **O(n) time**, but requires **O(n) space**; sorting can use less auxiliary space but takes **O(n log n) time**. Boyer–Moore satisfies both optimal targets simultaneously.
