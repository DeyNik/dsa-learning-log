### Find Pivot Index (Easy)

Given an array of integers `nums`, calculate the pivot index of this array.

The pivot index is the index where the sum of all the numbers strictly to the left of the index is equal to the sum of all the numbers strictly to the index's right.

If the index is on the left edge of the array, then the left sum is `0` because there are no elements to the left. This also applies to the right edge of the array.

Return the **leftmost pivot index**. If no such index exists, return `-1`.

---

### Examples

#### Example 1

* **Input:** `nums = [1, 7, 3, 6, 5, 6]`
* **Output:** `3`
* **Explanation:**
* The pivot index is `3`.
* Left sum = `nums[0] + nums[1] + nums[2] = 1 + 7 + 3 = 11`
* Right sum = `nums[4] + nums[5] = 5 + 6 = 11`



#### Example 2

* **Input:** `nums = [1, 2, 3]`
* **Output:** `-1`
* **Explanation:**
* There is no index that satisfies the conditions in the problem statement.



#### Example 3

* **Input:** `nums = [2, 1, -1]`
* **Output:** `0`
* **Explanation:**
* The pivot index is `0`.
* Left sum = `0` (no elements to the left of index `0`)
* Right sum = `nums[1] + nums[2] = 1 + (-1) = 0`



---

### Constraints

* $1 \le \text{nums.length} \le 10^4$
* $-1000 \le \text{nums[i]} \le 1000$

---

> **Note:** This question is the same as [LeetCode 1991: Find the Middle Index in Array](https://leetcode.com/problems/find-the-middle-index-in-array/).

````python
def pivotIndex(nums: list[int]) -> int:
        total = sum(nums)
        left = 0

        for i, x in enumerate(nums):
            right = total - left - x

            if left == right:
                return i

            left += x

        return -1

print(pivotIndex([1,7,3,6,5,6]))
print(pivotIndex([1,2,3]))
print(pivotIndex([2,1,-1]))
````

## Find Pivot Index

**Find Pivot Index** is LeetCode problem **724**, an Easy array problem centered on prefix sums. It asks for the leftmost index where the sum of every element strictly to its left equals the sum of every element strictly to its right, returning `-1` when no such index exists. [LeetCode](https://leetcode.com/problems/find-pivot-index/?utm_source=chatgpt.com)







### Core idea

The key observation is **`rightSum = totalSum - leftSum - nums[i]`**. First compute the entire array's sum, then scan from left to right while maintaining `leftSum`; the first position satisfying `leftSum == rightSum` is the required pivot. This avoids constructing separate prefix- and suffix-sum arrays.

### Example

For **`[1, 7, 3, 6, 5, 6]`**, index `3` is the pivot. The elements to its left sum to `1 + 7 + 3 = 11`, while those to its right sum to `5 + 6 = 11`; the pivot value `6` itself belongs to neither sum. [LeetCode](https://leetcode.com/problems/find-pivot-index/?utm_source=chatgpt.com)

### Efficient solution

A compact Python implementation is:

```python
class Solution:
    def pivotIndex(self, nums):
        total = sum(nums)
        left = 0

        for i, x in enumerate(nums):
            right = total - left - x

            if left == right:
                return i

            left += x

        return -1
```

This runs in **O(n) time** with **O(1) auxiliary space**: one pass obtains the total and another scans for the pivot. Scanning from left to right automatically ensures that the **leftmost** valid pivot is returned.

### Important edge cases

An endpoint can be a pivot because the sum on its empty side is defined as zero; for example, `[2, 1, -1]` has pivot index `0`. Negative numbers are also allowed, so approaches based on monotonic sums or two pointers are inappropriate. LeetCode notes that problem 724 is equivalent to **1991, Find the Middle Index in Array**.
