### Problem Description

Given an integer array `nums` sorted in non-decreasing order, remove the duplicates **in-place** such that each unique element appears only once. The relative order of the elements should be kept the same.

Consider the number of unique elements in `nums` to be `k`. After removing duplicates, return the number of unique elements `k`.

The first `k` elements of `nums` should contain the unique numbers in sorted order. The remaining elements beyond index `k - 1` can be ignored.

### Constraints

* $1 \le \text{nums.length} \le 3 \times 10^4$
* $-100 \le \text{nums}[i] \le 100$
* `nums` is sorted in non-decreasing order.

````python
def removeDuplicates(nums: list[int]) -> int:
        pt1=0
        pt2=1

        for _ in range(len(nums)):
            if pt2==len(nums):
                break

            if nums[pt1]==nums[pt2]:
                # print("pt has same value")
                # print("pt1",pt1,"pt2",pt2, "nums", nums)
                pt2+=1
                #print("pt1",pt1,"pt2",pt2, "nums", nums)
                continue

            #print(nums[pt1]<nums[pt1] , (pt2-pt1)>1)
            if nums[pt1]<nums[pt2] and (pt2-pt1)>1:
                # print("pointer have different values")
                # print("pt1",pt1,"pt2",pt2, "nums", nums)
                pt1+=1
                temp=nums[pt1]
                nums[pt1]=nums[pt2]
                nums[pt2]=temp
                #print("after swap:\n pt1",pt1,"pt2",pt2, "nums", nums)

                pt2+=1

                #print("pt1",pt1,"pt2",pt2, "nums", nums)

                continue

            pt1+=1
            pt2+=1
            # print("no condition")
            # print("pt1",pt1,"pt2",pt2)


        print(pt1+1)
        print(nums) 



removeDuplicates([1,1,2])
removeDuplicates([0,0,1,1,1,2,2,3,3,3,4])
removeDuplicates([1,2,3])
````

## Remove Duplicates from Sorted Array

*Remove Duplicates from Sorted Array* is **LeetCode Problem 26**, a classic array and two-pointer programming exercise. It is widely used to teach in-place array manipulation, efficient iteration, and interview problem-solving because it combines a simple premise with an important algorithmic pattern.







### Core idea

The problem asks for an algorithm that **removes duplicates from a sorted array in place** while using **constant extra memory**. Because the array is already sorted, equal values appear next to one another, allowing a single pass that keeps only the first occurrence of each distinct value.

### Common solution

The standard approach uses **two pointers**. One pointer tracks the position where the next unique value should be written, while the other scans through the array. Whenever a new value different from the previous unique value is found, it is copied forward and the write pointer advances. This preserves the relative order of the unique elements without allocating another array.

### Complexity

The optimal solution runs in **O(n)** time, where *n* is the length of the array, because every element is examined once. It requires **O(1)** auxiliary space since all updates occur directly within the original array.

### Why it matters

This problem is considered a foundational interview question because it reinforces several important concepts: exploiting **sorted input**, recognizing when **two-pointer techniques** simplify traversal, and performing **in-place modifications** under memory constraints. The same pattern appears in many later problems involving arrays, linked lists, sliding windows, and sequence processing.
