```python
def removeElement(nums: list[int], val: int) -> int:
        #similar to movezereos
        #two pointer
        left=0
        right=1
        count=0
        for i in range(len(nums)):
            #print("nums[",i,"]:",nums[i])
            print(left,right,count)

            if nums[i]==val:
                count+=1
            
            if right==len(nums):
                break
            
            if nums[left]==val and nums[right]!=val:
                temp=nums[left]
                nums[left]=nums[right]
                nums[right]=temp
            elif nums[left]==val and nums[right]==val:
                right+=1
                continue
           
            left+=1
            right+=1
        
        return left

nums1=[3,2,2,3]
answer=removeElement(nums1,3)
print(nums1, 3, "answer",answer, nums1[:answer] )
nums2=[0,1,2,2,3,0,4,2]
answer2=removeElement(nums2,2)
print(nums2,2,"answer:",answer2, nums2[:answer2])
nums3=[2]
answer3=removeElement([nums3],3)
print(nums3,3,"answer:",answer3, nums3[:answer3])
nums4=[3,3]
answer4=removeElement(nums4,5)
print(nums4,5,"answer:",answer4, nums4[:answer4])
```

```python
def sortedSquares(nums: list[int]) -> list[int]:
        #brute force
        for i in range(len(nums)):
            print("before",nums[i])
            nums[i] = nums[i]**2
            print("after",nums[i])
        
        print(nums)
        return sorted(nums)

sortedSquares([-4,-1,0,3,10])
```

```python
def intersection( nums1: list[int], nums2: list[int]) -> list[int]:
        intersection=[]
        for i in range(len(nums1)):
            if nums1[i] in nums2:
                intersection.append(nums1[i])
        
        return list(set(intersection))

print(intersection([1,2,2,1],[2,2]))
```

```python
def maxProduct(nums: list[int]) -> int:
        best=0
        curr=1
        for num in nums:
            curr=max(num, curr*num)
            best=max(best, curr)
            print(curr, best)

print(maxProduct([2,3,-2,4]))
print(maxProduct([-2,0,-1]))
```
