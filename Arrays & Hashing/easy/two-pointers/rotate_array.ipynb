

## Rotate Array

**Rotate Array** is LeetCode problem **189**, a Medium array problem that asks for an integer array to be rotated to the right by `k` positions. It is particularly useful for learning **in-place array manipulation**, because the follow-up asks for a solution using only `O(1)` extra space. [LeetCode](https://leetcode.com/problems/rotate-array/?envType=study-plan\&id=algorithm-i\&utm_source=chatgpt.com)







### The core problem

Given `nums = [1,2,3,4,5,6,7]` and `k = 3`, rotating right produces `[5,6,7,1,2,3,4]`. Since `k` can exceed the array length, the first useful normalization is **`k %= n`**; LeetCode currently allows arrays up to `10^5` elements and `k` up to `10^5`. [LeetCode](https://leetcode.com/problems/rotate-array/?envType=study-plan\&id=algorithm-i\&utm_source=chatgpt.com)

### The reversal trick

The standard optimal approach performs **three reversals**: reverse the entire array, reverse the first `k` elements, then reverse the remaining `n-k` elements. For `[1,2,3,4,5,6,7]`, `k=3` gives `[7,6,5,4,3,2,1] → [5,6,7,4,3,2,1] → [5,6,7,1,2,3,4]`. This achieves **`O(n)` time and `O(1)` extra space**. [LeetCode](https://leetcode.com/problems/rotate-array/?envType=study-plan\&id=algorithm-i\&utm_source=chatgpt.com)

### Why the trick works

Conceptually, split the original array into `A | B`, where `B` contains the final `k` elements. The target is `B | A`; reversing everything creates `reverse(B) | reverse(A)`, and separately reversing those two regions restores their internal order. This is the main insight that turns an apparently movement-heavy problem into a simple in-place transformation.

### Other approaches

A temporary array makes the indexing especially clear: an element originally at index `i` belongs at **`(i + k) % n`**, giving `O(n)` time but `O(n)` additional space. Another in-place technique follows **cyclic replacements**, carrying displaced values around index cycles; LeetCode explicitly points toward both extra-memory and cyclic approaches in addition to reversal. [LeetCode](https://leetcode.com/problems/rotate-array/?envType=study-plan\&id=algorithm-i\&utm_source=chatgpt.com)

### What it tests

The problem is less about rotation itself than recognizing **modular indexing, boundary handling, and space optimization**. Common mistakes include forgetting `k %= n`, reversing the wrong ranges, or implementing repeated one-position shifts—which can degrade to `O(nk)` time. The three-reversal solution is generally the cleanest interview solution because it combines linear runtime with constant auxiliary space.

```python
#O(n^2)
def rotate(nums: list[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        right=len(nums)-1

        for i in range(k):
            print(nums[right], nums[:right])
            nums=[nums[right]]+nums[:right]

        print(nums)

rotate([1,2,3,4,5,6,7],3)
        
```

```python
#O(n) but dont pass for larger values
def rotate(nums: list[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        nums[:]=nums[len(nums)- k:] + nums[:len(nums)-k]

        print(nums)

rotate([1,2,3,4,5,6,7],3)
        
```

```python
#O(n)
def rotate(nums: list[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        k%=len(nums)
        nums.reverse()

        nums[:k]=reversed(nums[:k])
        nums[k:]= reversed(nums[k:])

        print(nums)

rotate([1,2,3,4,5,6,7],3)
        
```
