### Problem Description

Given an integer array `nums`, move all `0`'s to the end of it while maintaining the relative order of the non-zero elements.

> **Note:** You must do this in-place without making a copy of the array.

---

### Examples

**Example 1:**

* **Input:** `nums = [0, 1, 0, 3, 12]`
* **Output:** `[1, 3, 12, 0, 0]`

**Example 2:**

* **Input:** `nums = [0]`
* **Output:** `[0]`

````python
def moveZeroes(nums: list[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        if len(nums)==1:
            pass
        pt1=0
        pt2=1
        for i in range(len(nums)):
            if pt2 == len(nums):
                break
            if nums[pt1]==0 and nums[pt2]==0:
                pt2+=1
                continue
            elif nums[pt1]==0 and nums[pt2]!=0:
                temp=nums[pt1]
                nums[pt1]=nums[pt2]
                nums[pt2]=temp
            pt1+=1
            pt2+=1
        
        print(nums)
            
````

### Problem requirements

The input is an integer array `nums`, which must be modified **in place**, without creating a separate result array. For example, `[0, 1, 0, 3, 12]` becomes `[1, 3, 12, 0, 0]`. The numbers `1`, `3`, and `12` retain their relative order, and the number of elements remains unchanged. [LeetCode](https://leetcode.com/problems/move-zeroes/description/?envId=admrlowh\&envType=problem-list-v2\&utm_source=chatgpt.com)

### How the two-pointer approach works

The standard solution uses a **read pointer** to inspect each element and a **write pointer** to track where the next nonzero element belongs. Whenever a nonzero value is found, it is moved into the next available position. A swap-based implementation leaves zeros behind as nonzero elements move forward, preserving their original order.

### Python solution

```python
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        left = 0

        for right in range(len(nums)):
            if nums[right] != 0:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
```

The `right` pointer scans the entire array, while `left` advances only when a nonzero element is encountered. An optional optimization skips swaps when `left == right`, avoiding unnecessary writes to elements already in their correct positions.

### Time and space complexity

The two-pointer solution has **O(n) time complexity**, because it examines each array element once, and **O(1) auxiliary space complexity**, because it uses only a fixed number of variables. These properties satisfy the problem's in-place requirement and make the approach efficient for arrays of up to 10,000 elements. ### Why this problem matters

Move Zeroes introduces **stable array partitioning**: separating elements according to a condition without changing the relative order of the retained elements. This technique also appears in array filtering, element removal, and data compaction. Its main lesson is that two coordinated pointers can reorganize an array efficiently without allocating additional storage.
