Given an array of integers `nums` and an integer `k`, return *the total number of subarrays whose sum equals to* `k`.

A subarray is a contiguous **non-empty** sequence of elements within an array.

### Example 1:
> **Input:** `nums = [1,1,1]`, `k = 2`  
> **Output:** `2`

### Example 2:
> **Input:** `nums = [1,2,3]`, `k = 3`  
> **Output:** `2`

### Constraints:
* $1 \le \text{nums.length} \le 2 \times 10^4$
* $-1000 \le \text{nums}[i] \le 1000$
* $-10^7 \le k \le 10^7$

```python
def subarraySum(nums: list[int], k: int) -> int:
    count = 0
    current_sum = 0
    # Maps {prefix_sum: number_of_times_seen}
    prefix_sums = {0: 1}  

    for num in nums:
        current_sum += num
        difference = current_sum - k

        # Add how many times we've seen the needed prefix sum
        if difference in prefix_sums:
            count += prefix_sums[difference]

        # Record the current running sum's frequency
        prefix_sums[current_sum] = prefix_sums.get(current_sum, 0) + 1

    return count

print(subarraySum([1,1,1],2))
print(subarraySum([1,2,3],3))
```

## Subarray Sum Equals K

*Subarray Sum Equals K* is a well-known array and prefix-sum programming problem from LeetCode (problem 560). It asks for the number of contiguous subarrays whose elements sum to a given value **k**, and is widely used to teach efficient use of **prefix sums** and **hash maps** in algorithm design.







### Core idea

The key insight is that if the **prefix sum** up to index *i* is `S`, then any earlier prefix sum equal to `S - k` identifies a contiguous subarray ending at *i* whose sum is exactly **k**. By storing how many times each prefix sum has appeared in a hash map, the algorithm can count all qualifying subarrays in a single pass instead of checking every possible range.

### Efficient solution

A brute-force approach examines many subarrays and typically requires **O(n²)** time. The standard optimal solution maintains a running prefix sum and a frequency map of previously seen prefix sums, achieving **O(n)** time and **O(n)** additional space. Initializing the map with `{0: 1}` is an important detail because it correctly counts subarrays that begin at index 0.

### Common pitfalls

Several implementation mistakes recur. The problem asks for the **number** of qualifying subarrays rather than their lengths or positions, duplicate prefix sums must increase a frequency count instead of replacing earlier values, and the frequency map should be updated **after** checking for `currentSum - k` so that only valid earlier prefixes are counted.

### Related concepts

Understanding this problem provides a foundation for many array and hashing techniques. Closely related ideas include **prefix sums**, **hash tables**, cumulative sums, and other subarray problems involving target sums, modular arithmetic, or fixed constraints, where recognizing relationships between prefix values often leads to linear-time algorithms.
