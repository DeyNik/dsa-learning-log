# Maximum Subarray

**Difficulty:** Medium

**Status:** Solved

**Topics / Companies**

Given an integer array `nums`, find the subarray with the largest sum, and return *its sum*.

---

### Example 1:

* **Input:** `nums = [-2,1,-3,4,-1,2,1,-5,4]`
* **Output:** `6`
* **Explanation:** The subarray `[4,-1,2,1]` has the largest sum `6`.

### Example 2:

* **Input:** `nums = [1]`
* **Output:** `1`
* **Explanation:** The subarray `[1]` has the largest sum `1`.

### Example 3:

* **Input:** `nums = [5,4,-1,7,8]`
* **Output:** `23`
* **Explanation:** The subarray `[5,4,-1,7,8]` has the largest sum `23`.

---

### Constraints:

* `1 <= nums.length <= 10^5`
* `-10^4 <= nums[i] <= 10^4`

---

> **Follow up:** If you have figured out the $\mathcal{O}(n)$ solution (Kadane's Algorithm), try coding another solution using the divide and conquer approach, which is more subtle.

````python
def maxSubArray(nums: list[int]) -> int:
        #Kadane's Algorithm
        current = best = nums[0]

        for x in nums[1:]:
            current = max(x, current+x)
            best = max(best, current)

        return best

print(maxSubArray([-2,1,-3,4,-1,2,1,-5,4]))
print(maxSubArray([1]))
print(maxSubArray([5,4,-1,7,8]))
````



## Maximum Subarray

Maximum Subarray is **LeetCode Problem 53**, a classic array and dynamic-programming problem: given an integer array, find the contiguous subarray with the largest possible sum and return that sum. It is especially notable for its elegant **O(n)** solution, commonly known as Kadane’s algorithm.

### Core idea

At each position, the key decision is whether to **extend the previous subarray** or **start a new subarray** at the current element. If `x` is the current value and `current` is the best sum ending immediately before it, the recurrence is **`current = max(x, current + x)`**; a separate `best` value tracks the largest `current` encountered anywhere.

### Example

For `[-2, 1, -3, 4, -1, 2, 1, -5, 4]`, the optimal contiguous subarray is **`[4, -1, 2, 1]`**, whose sum is **6**. Although including `-1` temporarily lowers the running sum, keeping it allows the later `2 + 1` to produce a better total than restarting.

### Kadane’s algorithm

A compact implementation keeps only two values, giving **O(n) time and O(1) extra space**:

```python
def maxSubArray(nums):
    current = best = nums[0]

    for x in nums[1:]:
        current = max(x, current + x)
        best = max(best, current)

    return best
```

Initializing from `nums[0]` is important because the array may contain **only negative numbers**. Initializing the answer to `0` would incorrectly allow an empty subarray.

### Dynamic-programming interpretation

Kadane’s algorithm is essentially **space-optimized dynamic programming**. Define `dp[i]` as the maximum sum of a nonempty subarray that **ends exactly at index `i`**; then `dp[i] = max(nums[i], dp[i-1] + nums[i])`, and the final answer is `max(dp)`. Since only `dp[i-1]` is needed, the entire DP array can be compressed into the single `current` variable above.
