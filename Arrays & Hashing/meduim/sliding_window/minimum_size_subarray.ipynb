## Minimum Size Subarray Sum

*Minimum Size Subarray Sum* is a well-known array and sliding-window programming problem, published as **LeetCode 209**. It asks for the length of the shortest contiguous subarray whose sum is at least a given target, making it a classic exercise for learning efficient two-pointer techniques instead of brute-force search.

### Problem

Given:
- An array of **positive integers** `nums`
- A positive integer `target`

The goal is to find the **minimum length** of a contiguous subarray whose sum is **greater than or equal to** `target`. If no such subarray exists, the answer is `0`.

For example:

- `target = 7`
- `nums = [2,3,1,2,4,3]`

The shortest qualifying subarray is `[4,3]`, so the answer is **2**.

### Key idea

The important observation is that **all numbers are positive**. This means:

- Expanding the window to the right can only increase the sum.
- Shrinking the window from the left can only decrease the sum.

Because of this monotonic behavior, a **sliding window (two pointers)** solution works efficiently.

The algorithm:
1. Extend the right end of the window while maintaining the running sum.
2. Whenever the sum reaches or exceeds `target`, record the current window length.
3. Shrink the window from the left as much as possible while keeping the sum at least `target`.
4. Continue until the end of the array.

### Complexity

The sliding-window solution is optimal for this problem.

| Approach | Time | Space |
|---|---:|---:|
| Brute force | **O(n²)** | O(1) |
| Sliding window | **O(n)** | O(1) |
| Prefix sums + binary search | **O(n log n)** | O(n) |

Each element enters and leaves the sliding window at most once, resulting in linear time.

### Why it's important

This problem is frequently used to teach:
- **Sliding window**
- **Two pointers**
- **Maintaining a running sum**
- Recognizing when the positivity of array elements enables linear-time algorithms

It is often one of the first problems programmers encounter when learning window-based techniques.

### Common pitfalls

A few implementation mistakes are common:

- Using the sliding-window approach on arrays that may contain **negative numbers**—the method is only correct because every element is positive.
- Forgetting to update the minimum length **before** shrinking the window.
- Returning the initial sentinel value instead of `0` when no qualifying subarray exists.
- Recomputing sums instead of maintaining a running total, which increases the time complexity.
