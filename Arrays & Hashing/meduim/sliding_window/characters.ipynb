## Longest Substring Without Repeating Characters

*Longest Substring Without Repeating Characters* is LeetCode Problem #3, a classic string-processing challenge that asks for the **length of the longest contiguous substring containing no duplicate characters**. It is widely used to teach the *sliding window* technique and hash-based lookups. [LeetCode](https://leetcode.com/problems/longest-substring-without-repeating-characters/?utm_source=chatgpt.com)

### Problem Statement

Given a string `s`, return the length of the longest substring without repeating characters. A **substring** must be contiguous.

Examples: [LeetCode](https://leetcode.com/problems/longest-substring-without-repeating-characters/?utm_source=chatgpt.com)

| Input | Output | Longest Valid Substring |
|---------|---------|---------|
| `"abcabcbb"` | `3` | `"abc"` |
| `"bbbbb"` | `1` | `"b"` |
| `"pwwkew"` | `3` | `"wke"` |

A common pitfall is confusing a **substring** with a **subsequence**. For example, `"pwke"` is not valid for `"pwwkew"` because its characters are not contiguous. [LeetCode](https://leetcode.com/problems/longest-substring-without-repeating-characters/?utm_source=chatgpt.com)

### Core Idea

The optimal solution uses a **sliding window**:

1. Maintain a window `[left, right]` containing unique characters.
2. Expand the window by moving `right`.
3. If a duplicate appears, move `left` until the duplicate is removed.
4. Track the maximum window size seen. [AlgoMonster](https://algo.monster/liteproblems/3?utm_source=chatgpt.com)

The key insight is that once a duplicate is found, there is no need to restart from scratch. The window can be adjusted incrementally, making the solution efficient. [AlgoMonster](https://algo.monster/liteproblems/3?utm_source=chatgpt.com)

### Optimal Solution (Python)

```python
def lengthOfLongestSubstring(s: str) -> int:
    seen = set()
    left = 0
    ans = 0

    for right in range(len(s)):
        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])
        ans = max(ans, right - left + 1)

    return ans
```

### Complexity

The optimal sliding-window approach runs in:

- **Time:** `O(n)`
- **Space:** `O(min(n, alphabet_size))`

Each character enters and leaves the window at most once. [AlgoMonster](https://algo.monster/liteproblems/3?utm_source=chatgpt.com)
