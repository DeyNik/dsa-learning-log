## A Python Built-in Type
Among the principal built-in types - numerics, sequences, mappings, classes, instances and exceptions, list is a sequence type. Other sequence type include - tuple and range.


## Characteristics:
- Ordered (Insertion sequence is preserved)
- Mutable
- Indexed
- Iterable
- Heterogeneous — can contain different types
- Dynamically sized
- Allows duplicate values
- Can contain other lists
- Can contain any Python object


A list stores references to objects in a specific order.




##### **List are not hashable as they are mutable, so they cannot be even used as dictionary keys.**

### Creating List

#### uses class list(iterable=(), /)¶
Lists may be constructed in several ways:

- Using a pair of square brackets to denote the empty list: []

- Using square brackets, separating items with commas: [a], [a, b, c]

- Using a list comprehension: [x for x in iterable]

- Using the type constructor: list() or list(iterable)

>The constructor builds a list whose items are the same and in the same order as iterable’s items. iterable may be either a sequence, a container that supports iteration, or an iterator object. If iterable is already a list, a copy is made and returned, similar to iterable[:]. For example, list('abc') returns ['a', 'b', 'c'] and list( (1, 2, 3) ) returns [1, 2, 3]. If no argument is given, the constructor creates a new empty list, [].


````python
a = []

b = list()

c = [1, 2, 3]

numbers = [10, 20, 30]

mixed = [10, "hello", 3.14, True, None]

nested = [[1, 2], [3, 4]]

duplicates = [1, 1, 2, 2, 3]

#From another iterable
list("hello")
# ['h', 'e', 'l', 'l', 'o']
list((1, 2, 3))
# [1, 2, 3]
list({1, 2, 3})
# [1, 2, 3]   # order shouldn't be relied upon from a set
list(range(5))
# [0, 1, 2, 3, 4]

#list() consumes an iterable.

list(x for x in range(5))

[0, 1, 2, 3, 4]
````

### Indexing
Zero based indexing

````python
a= [1,2,3, 7, 9,3]
print("Original List",a)
print("Indexing")
print("a[0]: ",a[0])
print("a[0]: ",a[-0])

# index out of range print(a[9])

print("Length:" , len(a))

#get last element 
print("Last Element: a[-1]: ",a[-1]," a[len(a)-1]:", a[len(a)-1])

# [start, stop-1, step]
print("Slicing")
print("a[1:3]: ",a[1:3])
print("a[::-1]: ",a[::-1]) # reverse the list
print("a[-3:]: ",a[-3:]) #negative slicing
print("a[::2]: ",a[::2]) #step
print("a[5:1:-2]: ",a[5:1:-2]) #negative step

print("Slice Assignment")
a[1:3]=[4,5]
print("a[1:3]=[4,5]: ",a)
a[::2]=[10,20,30] #same length assigment always
print("a[::2]=[10,20,30]: ",a)
````

### Repetiton
**s*n** or **n*s** or 

````python
ls=[[]]*3
print(ls)
ls[0].append(15)
print(ls)
````

#### What Happened?
list = [[]] * 3 creates one inner list object and stores three references to that same object.

```mermaid
flowchart LR
    A["list = [[]]"] --> B["list"]
    B --> C["0"]
    B --> D["1"]
    B --> E["2"]

    C --> F["same [] object"]
    D --> F
    E --> F
```

Thus making any change, results in

```mermaid
flowchart LR
    A["list"] --> B["0"]
    A --> C["1"]
    A --> D["2"]

    B --> E["[15]"]
    C --> E
    D --> E
```

**Do this instead for a multidimensional array**

````python
ls = [[] for i in range(3)]
ls[0].append(3)
print(ls)
````

### Shallow vs Deep Copy

````python
a=[[2,3],[3,4]]
b=a.copy()
# b=a[:]
# b=list(a)

b[0].append(35)
print("Shallow Copy")
print("a: ",a)
print("b: ",b)

print("Deep Copy")
import copy
b_new=copy.deepcopy(a)
b_new[0].append(35)
print("a: ",a)
print("b_new: ",b_new)
````

### Python passes object references

- This is sometimes described as: **Call by object sharing.**

- Lists are mutable, so a function can mutate the object passed to it.

- But:
```
def change(items):
    items = [99]
```

- doesn't replace the caller's variable.

- This distinction between mutation and rebinding is essential.

### Important Gotchas

````python
a=[4,5,6,7,7,7]

# 1. append = one item
a.append([1, 2])
# [ ..., [1, 2] ]

# 2. extend = many items
a.extend([1, 2])
# [..., 1, 2]

# 3. sort modifies and returns None
a.sort()

# 4. sorted creates a new list
b = sorted(a)

# 5. reverse modifies and returns None
a.reverse()

# 6. slicing creates a new outer list
b = a[:]

# 7. assignment does NOT copy
b = a

# 8. remove = value
a.remove(10) #only remove the first element

# 9. pop = index + returns item
x = a.pop(0)

# 10. del = index/slice, doesn't return item
del a[0]

# 11. [] * n can share nested mutable objects
a = [[] for _ in range(n)]

# 12. lists are mutable and unhashable
````
