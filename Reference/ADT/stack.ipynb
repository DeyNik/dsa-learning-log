## Core Operations
| Operation      | Meaning                      | Typical complexity |
| -------------- | ---------------------------- | -----------------: |
| `push`         | Add item to top              |             `O(1)` |
| `pop`          | Remove top item              |             `O(1)` |
| `peek` / `top` | Look at top without removing |             `O(1)` |
| `is_empty`     | Check whether stack is empty |             `O(1)` |
| `size`         | Number of items              |             `O(1)` |


#### Push

```python
stack=[]

#push
stack.append(10)
stack.append(20)
stack.append(30)

print(stack)
```

##### Pop

```python
x= stack.pop()
print(x)
print(stack)
```

#### Peek

```python
top= stack[-1]
print(top)
```

#### Check Empty

```python
if not stack:
    print("Satck is empty")
```

#### Size

```python
len(stack)
```

### Stacks with collections.deque

|                      | `list`     | `deque`    |
| -------------------- | ---------- | ---------- |
| Push                 | `append()` | `append()` |
| Pop                  | `pop()`    | `pop()`    |
| Peek                 | `[-1]`     | `[-1]`     |
| Right-end operations | **O(1)**   | **O(1)**   |
| Left-end operations  | O(n)       | **O(1)**   |
| Typical stack choice | Yes      | Yes      |


```python
from collections import deque
stack = deque()

stack.append(10)
stack.append(20)

stack.pop()
stack[-1]
```

| Operation            |         Complexity |
| -------------------- | -----------------: |
| `append(x)`          | **O(1) amortized** |
| `pop()`              |           **O(1)** |
| `stack[-1]`          |           **O(1)** |
| `len(stack)`         |           **O(1)** |
| `stack.insert(0, x)` |               O(n) |
| `stack.pop(0)`       |               O(n) |
| Search `x in stack`  |               O(n) |
| Access `stack[i]`    |               O(1) |
