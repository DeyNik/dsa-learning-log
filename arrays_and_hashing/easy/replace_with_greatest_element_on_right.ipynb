## Replace Elements with Greatest Element on Right Side - 1299

Given an array arr, replace every element in that array with the greatest element among the elements to its right, and replace the last element with -1.

After doing so, return the array.

 
```text
Example 1:

Input: arr = [17,18,5,4,6,1]
Output: [18,6,6,6,1,-1]
Explanation: 
- index 0 --> the greatest element to the right of index 0 is index 1 (18).
- index 1 --> the greatest element to the right of index 1 is index 4 (6).
- index 2 --> the greatest element to the right of index 2 is index 4 (6).
- index 3 --> the greatest element to the right of index 3 is index 4 (6).
- index 4 --> the greatest element to the right of index 4 is index 5 (1).
- index 5 --> there are no elements to the right of index 5, so we put -1.
Example 2:

Input: arr = [400]
Output: [-1]
Explanation: There are no elements to the right of index 0.
```

### brute force

````python
from typing import List


def largest_element(arr: List[int]) -> int:
    largest_number=float('-inf')
    largest_index=0

    for i in range(0,len(arr)):
        if arr[i]>largest_number:
            largest_number=arr[i]
            largest_index=i
    
    return largest_index


class Solution1:
    
    def replaceElements(self, arr: List[int]) -> List[int]:
        result= [] 

        if len(arr)==1:
            return [-1]

        for i in range(len(arr)-1):
            idx=largest_element(arr[i+1:])
            actual_idx=i+1+idx
            result.append(arr[actual_idx])

        result.append(-1)
        
        return result
            
````

````python
class Solution2:
    
    def replaceElements(self, arr: List[int]) -> List[int]:
        result= [] 

        if len(arr)==1:
            return [-1]

        for i in range(len(arr)-1):
            result.append(max(arr[i+1:]))

        result.append(-1)
        
        return result
            
````

### Optimal

````python
class Solution3:
    
    def replaceElements(self, arr: List[int]) -> List[int]:
        rightmax=-1

        for i in range(len(arr)-1,-1,-1):
            newmax=max(rightmax, arr[i])
            arr[i]=rightmax
            rightmax=newmax

        return arr
            
````

````python
#test cases

s=Solution3()#Solution1()

#true cases-----------------------------
print(s.replaceElements([17,18,5,4,6,1])) #[18,6,6,6,1,-1]
print(s.replaceElements([400])) #[-1]

````

### Example:
**nums=[17,18,5,4,6,1]**

As we can clearly see from the problem statement, the solution should be formed by replacing the element in the array with the largest number in its right, we would try solving in the same lines.

We have three options:
1. Brute Force: For each element , find the largest number in the right and replace. It would take O(n^2) time and your leetcode test cases would fail for exceeding the time limit, like mine. But at least you grasped the intuition. 

2. Now just like I did, you would want to replace the method for finding the largest number with **max**. It still takes O(n^2) but, helps a lot. Would it save you from exceeding time limits? Nope, however, you're very close to the actual solution.
    

3. The OG neetcode breaks it down beautifully. Assume you want to pick the largest number at your right.
    So , in this case:
    [17,18,5,4,6,1] -> from third last to 18, the largest number to its right is 6?
    - That is, for the third last number or 4, **the largest is the  6**
    - for the fourth last or 5, it is **4 or the largest number for third last**
    - for the fifth, or 18 , it is **5 or the largest number for the fourth last**
    - and so on...

        hence, its max(arr[i],lastMax)
    
    Time Complexity ~ O(n) as max is now only comparing 2 elements which is equivalent to a if (a > b else b). You can alternatively use that in your code.
