## valid Word Square - 888

Given a sequence of words, check whether it forms a valid word square.

A sequence of words forms a valid word square if the k^th row and column read the exact same string, where 0 ≤ k < max(numRows, numColumns).

- The number of words given is at least 1 and does not exceed 500.
- Word length will be at least 1 and does not exceed 500.
- Each word contains only lowercase English alphabet a-z.


### Problem Link
[valid word square - Lint Code](https://www.lintcode.com/problem/888/)


```python
class Solution:
    """
    @param words: a list of string
    @return: a boolean
    """
    def valid_word_square(self, words):
        # Write your code here
        for i in range(len(words)):
            for j in range(len(words[i])):
                if j>=len(words) or i>=len(words[j]):
                    return False
                if words[i][j]!=words[j][i]:
                    return False
        return True
        
```

```python
#test cases

s=Solution()#Solution1()

'''
  "a b c d",
  "b n r t",
  "c r m y",
  "d t y e"
'''
print(s.valid_word_square([
  "abcd",
  "bnrt",
  "crmy",
  "dtye"
])) #true

'''
  "b a l l",
  "a r e a",
  "r e a d",
  "l a d y"
'''
print(s.valid_word_square([
  "ball",
  "area",
  "read",
  "lady"
]))#false

'''
"b a l l",
"a s e e",
"l e t",
"l e p"
'''
print(s.valid_word_square(["ball","asee","let","lep"]))
```
