# Binary Search

- List is sorted
- O(logn) where we use log base 2


```python
def binary_search(arr:list[int], item:int)->int|None:
    #define beginning and end
    beg=0
    end=len(arr)-1

    #loop the items in array to find the element
    while beg<=end:
        mid=(beg+end) # calculate mid at all times  to ensure search space reduces
    
        if item>arr[mid]:
            beg=mid +1
        elif item<arr[mid]:
            end=mid-1
        elif item==arr[mid] :
            return mid

    return None

arr=[2,5,5,6,7,8,27]
result=binary_search(arr,8)

if result is None: 
    print("Element not found")
else:
    print("Element found at:", result)
```

```python
def findSmallest(arr):
    smallest_index:int=0
    for i in range(1,len(arr)):
        if arr[i]<arr[smallest_index]:
            smallest_index=i
    return smallest_index

def selectionSort(arr):
    newArr:list[int]=[]
    for i in range(0,len(arr)):
        smallest_index= findSmallest(arr)
        newArr.append(arr.pop(smallest_index))
    return newArr

print(selectionSort([5,4,2,1,67,0]))
```

```python
def factorial(n):
    if n==0 or n==1:
        return n

    return n*factorial(n-1)

print(factorial(5))
```

# Divide and Conquer
- recursive algorithms
- Two step solution:
    - figure out base case(simplest case)
    - divide aor decrease your problem until it becomes the base case

```python

def sum_of_numbers(n:list[int]):
    #base case
    if len(n)==0:
        return 0

    return n[0]+ sum_of_numbers(n[1:])

print(sum_of_numbers([2,4,6]))
```

### TIP: 
When you’re writing a recursive function involving an array, the base case is 
often an empty array or an array with one element. If you’re stuck, try that irst.

```python
#4.3
def global_maxima(n):
    if len(n)==2:
        return n[0] if n[0]>n[1] else n[1]

    local_maxima = global_maxima(n[1:])
    return n[0] if n[0]> local_maxima else local_maxima

print(global_maxima([2,3,45,3,6,56,6]))
```

```python
#4.4
def recursive_binary_search(arr, item):
    if len(arr)==0:
        return None 
    
    mid=len(arr)//2
    
    if arr[mid]==item:
        return mid
    elif arr[mid]>item:
        return recursive_binary_search(arr[:mid],item)
    elif arr[mid]<item:
        return recursive_binary_search(arr[mid+1:], item)

recursive_binary_search([3,5,8,9],8)
```

### QUICKSORT

```python
def quick_sort(arr):
    if len(arr)<2:
        return arr
    
    pivot=arr[0]
    smaller_than_pivot=[i for i in arr[1:] if i<=pivot]
    larger_than_pivot=[i for i in arr[1:] if i>pivot]
    
    return quick_sort(smaller_than_pivot)+ [pivot] + quick_sort(larger_than_pivot)

quick_sort([34,5,2,5,2,1])
```

# Breadth First Search
Algorithm to solve shortest path problem

### What is a graph?
- Set of connections
- contains node and edge
- first degree aka closest connections are traversed first. Only works if search is in the same oredr that they are added, which means, using a queue
- Use Hash Tables to store nodes and edges like graph["A"] = ["B","C","D"]
#### Queue
- Enqueue(Add an item) and Dequeue(remove an item)
- First In, First Out

### Directed Graph 
Relationship is only one way

### Undirected Graph
Relationship is both ways. Has no arrows. Both are each others neighbours.Hence it becomes a cycle

```python
graph={}
graph["you"]=["alice","bob","claire"]
graph["bob"]=["anuj","peggy"]
graph["alice"]=["peggy"]
graph["claire"]=["thom","jonny"]
graph["anuj"]=[]
graph["peggy"]=[]
graph["thom"]=[]
graph["jonny"]=[]

print(graph)

from collections import deque
search_queue=deque()
search_queue += graph["you"]

def person_is_seller(name):
    return name[-1] == 'm'

def search_mango_seller():
    global search_queue
    while search_queue:
        person= search_queue.popleft()
        if person_is_seller(person):
            print(person , "is a mango seller.")
            return True
        else:
            search_queue+=graph[person]
    return False 

search_mango_seller()
```

```python
#to avoid infinte loop we keep a check of nodes already traversed

def search(name):
    search_queue = deque()
    search_queue+=graph[name]
    searched=[]

    while search_queue:
        person = search_queue.popleft()

        if not person in searched:
            if person_is_seller(person):
                print(person, "is a mango seller")
                return True
            else:
                search_queue += graph[person]
                searched.append(person)
    return False

search("claire")
```
